```python
import heapq
from typing import List, Dict, Tuple

SMALL_FIT_PENALTY = 8
LARGE_FIT_PENALTY = 2

MEASUREMENT_WEIGHTS = {
    "Chest": 2,
    "Waist": 1,
    "Hip": 1
}
```

```python
garments = [

    {
        "id": "G101",
        "Chest": 100,
        "Waist": 84,
        "Hip": 98
    },

    {
        "id": "G102",
        "Chest": 96,
        "Waist": 82,
        "Hip": 96
    },

    {
        "id": "G103",
        "Chest": 104,
        "Waist": 88,
        "Hip": 102
    },

    {
        "id": "G104",
        "Chest": 98,
        "Waist": 85,
        "Hip": 97
    }
]

user = {

    "Chest": 98,
    "Waist": 84,
    "Hip": 99

}
```

```python
def calculate_fit_score(
    user: Dict[str, float],
    garment: Dict[str, float]
) -> float:

    score = 100.0

    for measurement, weight in (
        MEASUREMENT_WEIGHTS.items()
    ):

        user_value = user[measurement]
        garment_value = garment[measurement]

        difference = garment_value - user_value

        if difference < 0:

            penalty = (
                abs(difference)
                * SMALL_FIT_PENALTY
            )

        else:

            penalty = (
                difference
                * LARGE_FIT_PENALTY
            )

        score -= penalty * weight

    score = max(min(score, 100), 0)

    return round(score, 2)
```

```python
def find_best_fits(
    garments: List[Dict],
    user: Dict[str, float],
    top_n: int = 3
) -> List[Tuple[float, str]]:

    results = []

    for garment in garments:

        fit_score = calculate_fit_score(
            user,
            garment
        )

        results.append(
            (
                fit_score,
                garment["id"]
            )
        )

    best_matches = heapq.nlargest(
        top_n,
        results
    )

    return best_matches
```

```python
if __name__ == "__main__":

    try:

        top3 = find_best_fits(
            garments,
            user
        )

        print("\nTop 3 Best-Fit Garments:\n")

        for score, garment_id in top3:

            print(
                f"Garment ID: {garment_id} | "
                f"Fit Confidence Score: "
                f"{score:.2f}"
            )

    except Exception as error:

        print(
            f"Error occurred: {error}"
        )
```
